Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool, 
built on top of the Python programming language. It allows you to work with labeled and relational data in an intuitive way.

```python
import pandas as pd # import pandas as pd
pd.set_option('display.width', 1000) # This is to avoid line breaks in the output of the dataframe
```

```python
# Loading data from a CSV file.
df = pd.read_csv('pokemon_data.csv')
```

```python
# Print the first 5 rows of the dataframe to get an overview of the data.
print(df.head(5)) # Display the first 5 rows of the dataframe.
```

```python
print(df.tail(5)) # Display the last 5 rows of the dataframe.
```

```python
# Reading columns from the dataframe.
print(df.columns) # Display the column names of the dataframe.
```

```python
# Displaying specific columns from the dataframe.
print(df['Name']) # Display the 'Name' column of the dataframe.
```

```python
# Display multiple columns from the dataframe.
print(df[['Name', 'Type 1', 'HP']]) # Display the 'Name', 'Type 1', and 'HP' columns of the dataframe.
```

```python
# Display each row of the dataframe using iloc.
print(df.iloc[0:4]) # Display the first 4 rows of the dataframe using iloc.
```

```python
# Read A specific location (Row 2, Column 1) (Name of the 3rd Pokemon)
print(df.iloc[2, 1]) # Display the value at row index 2 and column index 1 of the dataframe.
```

```python
# Display a filtered column based on a condition.
print(df.loc[df['Type 1'] == 'Fire']) # Display all rows where the 'Type 1' column is equal to 'Fire'.
```

```python
# Describing Data
df.describe() # Display summary statistics of the dataframe.
```

```python
# Sorting Data
df.sort_values('Name', ascending=False)
```

```python
# Sorting by multiple columns
df.sort_values(['Type 1', 'Name'], ascending=[False, True]) # Sort by 'Type 1' and then by 'Name' in ascending order.
```

```python
# Making changes to the data
print(df.head(5))
```

```python
# Adding a new column to the dataframe
df['Total'] = df['HP'] + df['Attack'] + df['Defense'] + df['Sp. Atk'] + df['Sp. Def'] + df['Speed']

print(df.head(5))
```

```python
# Dropping a column
df = df.drop(columns=['Total']) # Have to store this in a var because drop() does not modify the dataframe in place, meaning the original dataframe remains unchanged unless reassigned.

print(df.head(5))
```

```python
# Adding Total column via iloc
df['Total'] = df.iloc[:, 4:10].sum(axis=1) # axis = 1 means summing across columns for each row instead of down the columns

print(df.head(5))
```

```python
# Move a column to a different position
col = df.pop('Total') # Remove the column from its current position
df.insert(4, 'Total', col) # Insert the column at the desired position (index 4 in this case)

print(df.head(5))
```

```python
# Saving our data
# df.to_csv('modified_pokemon_data.csv', index=False) # Save the dataframe to a CSV file without the index column
```

```python
# Filtering data
filtered_total = df[df['Total'] > 700] # Example: filter rows where the 'Total' column is greater than 500

print(filtered_total)
```

```python
# Filtering data on multiple conditions
filtered_multiple = df[(df['Total'] > 700) & (df['HP'] > 105)] # Example: filter rows where 'Total' is greater than 700 and 'HP' is greater than 105
filtered_multiple = filtered_multiple.reset_index(drop=True) # Drop = True means the old index is not added as a column

print(filtered_multiple)
```

```python
# Regex filtering
regex_filtered = df[df['Type 1'].str.contains('Fire|Water', regex=True)] 

print(regex_filtered)
```

```python
# Conditional changes
# Changing 'fire' to 'flamer'
df.loc[df['Type 1'] == 'Fire', 'Type 1'] = 'Flamer'

print(df.head(5))
```

```python
# Aggregate statistics
mean_total = df['Total'].mean() # Calculate the mean of the 'Total' column
max_total = df['Total'].max() # Calculate the maximum of the 'Total' column
min_total = df['Total'].min() # Calculate the minimum of the 'Total' column
sum_total = df['Total'].sum() # Calculate the sum of the 'Total' column

print(f"Mean Total: {mean_total}")
print(f"Max Total: {max_total}")
print(f"Min Total: {min_total}")
print(f"Sum Total: {sum_total}")
```
